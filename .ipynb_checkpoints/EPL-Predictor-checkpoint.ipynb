```python
import pandas as pd
from sklearn import tree
from sklearn.ensemble import RandomForestClassifier
from sklearn import cross_validation
from xgboost import XGBClassifier

import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = pd.read_csv('final_dataset1.csv')

data = data[data.MW > 3]

data.drop(['Unnamed: 0', 'HomeTeam', 'AwayTeam', 'HM1', 'HM2', 'HM3', 'AM1', 'AM2', 'AM3', 'Date', 'MW', 'HTFormPtsStr', 'ATFormPtsStr', 'FTHG', 'FTAG',
           'HTGS', 'ATGS', 'HTGC', 'ATGC','DiffLP','DiffPts','DiffFormPts',
           'HM4','HM5','AM4','AM5','HTLossStreak5','ATLossStreak5','HTWinStreak5','ATWinStreak5',
           'HTWinStreak3','HTLossStreak3','ATWinStreak3','ATLossStreak3'],1, inplace=True)

display(data.head())
```

```python
data.shape[1]
```

```python
n_matches = data.shape[0]

n_features = data.shape[1] - 1


n_homewins = len(data[data.FTR == 'W'])

win_rate = (float(n_homewins) / (n_matches)) * 100

# Print the results
print ("Total number of matches: {}".format(n_matches))
print ("Number of features: {}".format(n_features))
print ("Number of matches won by home team: {}".format(n_homewins))
print ("Win rate of home team: {:.2f}%".format(win_rate))
```

```python
X_all = data.drop(['FTR'],1)
y_all = data['FTR']


from sklearn.preprocessing import scale

cols = [['HTGD','ATGD','HTP','ATP','HomeTeamLP','AwayTeamLP']]
for col in cols:
    X_all[col] = scale(X_all[col])
X_all
```

```python
from sklearn.cross_validation import train_test_split

# Shuffle and split the dataset into training and testing set.
X_train, X_test, y_train, y_test = train_test_split(X_all, y_all, 
                                                    test_size = .2,
                                                    random_state = 12)
```

```python
X_train.shape[1]
```

```python
modelLR = XGBClassifier()
modelLR.fit(X_train,y_train)
```

```python
modelRandom = RandomForestClassifier(n_estimators = 20, criterion = 'entropy', random_state = 42)
modelRandom.fit(X_train,y_train)
```

```python
randomPredictions = modelRandom.predict(X_test)
display(randomPredictions)
```

```python
predictionsLR = modelLR.predict(X_test)
display(predictionsLR)
```

```python
from sklearn.metrics import accuracy_score
print('Accuracy score:', accuracy_score(y_test,randomPredictions))
print('Accuracy score:', accuracy_score(y_test,predictionsLR))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,predictionsLR))
```

```python
drone= pd.read_csv('final_dataset1.csv')
ht="Man United"
at="Man City"
```

```python
a = drone["HomeTeam"]
count = 0
for i in a:
    if i == ht:
        count= count+1

b= drone["AwayTeam"]
count2 = 0
for i in b:
    if i == at:
        count2= count2+1

count,count2
```

```python
a= (drone[drone["HomeTeam"] == ht].sum().HTP)/count
b= (drone[drone["AwayTeam"] == at].sum().ATP)/count2
c= (drone[drone["HomeTeam"] == ht].sum().HomeTeamLP)/count
d= (drone[drone["AwayTeam"] == at].sum().AwayTeamLP)/count2
e= (drone[drone["HomeTeam"] == ht].sum().HTFormPts)/count
f= (drone[drone["AwayTeam"] == at].sum().ATFormPts)/count2
g= (drone[drone["HomeTeam"] == ht].sum().HTGD)/count
h= (drone[drone["AwayTeam"] == at].sum().ATGD)/count2
```

```python
p=[a,b,c,d,e,f,g,h]
print(p)
len(p)
```

```python
df = pd.DataFrame({'HTP': p[0],'ATP': p[1],'HomeTeamLP': p[2],'AwayTeamLP': p[3],'HTFormPts': p[4],'ATFormPts': p[5],
                   'HTGD': p[6],'ATGD': p[7]}, index=[0])
df
```

```python
result= modelLR.predict(df)
```
